```python
from google import genai

# Inicializa el cliente; buscará automáticamente tu variable de entorno GEMINI_API_KEY
client = genai.Client(api_key = api_key)

# Genera contenido utilizando el modelo recomendado
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Cuando son las matriculas?",
)

print(response.text)
```
