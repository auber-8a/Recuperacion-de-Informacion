```python
# Importar librerias
import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('stopwords')
nltk.download('wordnet')
```

```python
# Cargar el Dataset
movies = pd.read_csv("data//rotten_tomatoes_movies.csv")
reviews = pd.read_csv("data//rotten_tomatoes_critic_reviews.csv")

print(movies.shape)
print(reviews.shape)
```

```python
# Union de los 2 datasets para trabajar mas facil
df = pd.merge(
    reviews,
    movies,
    on='rotten_tomatoes_link',
    how='left'
)
```

```python
df.head()
```

```python
df.columns
```

```python
df['text'] = (
    df['movie_title'].fillna('') + ' ' +
    df['genres'].fillna('') + ' ' +
    df['critics_consensus'].fillna('') + ' ' +
    df['review_content'].fillna('')
)
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):

    # minúsculas
    text = text.lower()

    # eliminar puntuación
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    # eliminar espacios extra
    text = re.sub(r'\s+', ' ', text).strip()

    # tokenización
    words = text.split()

    # eliminar stopwords + lematización
    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]

    return ' '.join(words)
```

```python
df['processed_text'] = df['text'].apply(preprocess)
```

# Parte 2 — Generación de embeddings

Para representar los documentos se utilizó el modelo
'all-MiniLM-L6-v2' de Sentence Transformers.

Este modelo genera representaciones vectoriales densas capaces de capturar relaciones semánticas entre documentos y consultas.

Cada documento fue transformado en un embedding de alta dimensionalidad, permitiendo posteriormente calcular similitud semántica mediante similitud coseno.

```python
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

Reducí el tamaño ya que era gigantesco y en las compus del lab se demoraba demasiado.  
Lo reducí a 5000.

```python
df = df.sample(5000, random_state=42)
```

```python
document_embeddings = model.encode(
    df['processed_text'].tolist(),
    show_progress_bar=True
)
```

```python
document_embeddings.shape
```

```python
import numpy as np

np.save("document_embeddings.npy", document_embeddings)
```

# Parte 3

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def search(query, top_k=5):

    processed_query = preprocess(query)

    query_embedding = model.encode([processed_query])

    similarities = cosine_similarity(
        query_embedding,
        document_embeddings
    )[0]

    top_indices = similarities.argsort()[::-1][:top_k]

    results = []

    for rank, idx in enumerate(top_indices, start=1):

        review_text = str(df.iloc[idx]['review_content'])

        results.append({
            'Ranking': rank,
            'Document_ID': idx,
            'Movie_Title': df.iloc[idx]['movie_title'],
            'Similarity': round(similarities[idx], 4),
            'Review_Fragment': review_text[:200]
        })

    return pd.DataFrame(results)
```

```python
query = "science fiction movie with advanced technology"

search(query)
```

```python
from google import genai
from dotenv import load_dotenv
import os

# Cargar las variables de entorno desde el archivo .env oculto
load_dotenv()

# Inicializar el cliente oficial de Gemini (toma automáticamente la variable GEMINI_API_KEY)
client = genai.Client()
```

```python
def sistema_recuperacion_informacion(query, top_k=3):
    # 1. Preprocesar y vectorizar la consulta del usuario usando las funciones existentes
    processed_query = preprocess(query)
    query_embedding = model.encode([processed_query])

    # 2. Calcular las similitudes con la matriz document_embeddings
    similarities = cosine_similarity(query_embedding, document_embeddings)[0]
    top_indices = similarities.argsort()[::-1][:top_k]
    
    # 3. Concatenar los resultados en una cadena limpia de texto
    fragmentos_encontrados = []
    for idx in top_indices:
        titulo = df.iloc[idx]['movie_title']
        consenso_o_resena = str(df.iloc[idx]['review_content'])
        
        item = f"Película: {titulo} | Reseña: '{consenso_o_resena}'"
        fragmentos_encontrados.append(item)
        
    resultados = "\n".join(fragmentos_encontrados)
    return resultados
```

```python
def ejecutar_chatbot(query):
    # Obtener lo que encontró tu Sistema de Recuperación de Información
    resultados = sistema_recuperacion_informacion(query, top_k=3)
    
    # Construcción exacta del prompt 
    prompt = f"""
Mi usuario quiere saber ({query}) y mi Sistema de Recuperacion de Informacion encontro ({resultados}). Dame la respuesta en base al contexto dado por el Sistema de Recuperacion de Informacion.
"""

    # Enviar el prompt estructurado al modelo recomendado gemini-2.5-flash
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
    )
    
    return response.text
```
